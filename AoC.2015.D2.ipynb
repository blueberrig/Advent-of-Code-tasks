```python
#AoC 2015 Day 1 Part 1
with open('input.txt', 'r') as file:
    all_lines = [line.strip() for line in file]

data = [[int(n) for n in line.strip().split('x')] for line in all_lines]
slack=0;
paper=0;
for i in range(len(data)):
    l,w,h=data[i]
    slack=min(l*w,w*h,h*l)
    paper=paper+2*l*w+2*w*h+2*h*l+slack
print("Total square feet of wrapping paper needed:",paper)
#Part 2
bow=0;
ribbon=0;
for i in range(len(data)):
    l,w,h=data[i]
    bow=l*w*h
    ribbon=ribbon+min(2*(l+w),2*(w+h),2*(h+l))+bow
print("Total feet of ribbon needed:",ribbon)
```
